```python
#Resample the data by month and plot the monthly average temperature over time.
#firstly coverting index to datetime indexabs
df['DATE'] = pd.to_datetime(df['DATE'])
df.set_index('DATE',inplace=True)
#mothly_temp=df.resample('M')['temp'].mean()
```

```python
#Calculate the rolling 7-day average of temp and add it as a new column.
df['temp_rolling7']=df['temp'].rolling(7).mean()
df
```

```python
#Find the hottest day and the coldest day in the entire dataset.
hottest=df.loc[df['temp'].idxmax()]
coldest=df.loc[df['temp'].idxmin()]
print('hottest', hottest['temp'])

print('coldest', coldest['temp'])
```

```python
#What is the average temperature difference between day and night? Use tempmax and tempmin.
df['temp_diff']=df['tempmax']-df['tempmin']
print(f"the average day and night temperature differnce is {df['temp_diff']}")
```

```python
#Calculate the correlation between humidity and feelslike temperature.
corr_humidity_feels=df['humidity'].corr(df['feelslike'])
print(f"the realtion between humidity and feelslike is {corr_humidity_feels}")
```

```python
#Calculate the median, standard deviation, and interquartile range (IQR) for temperature
temp_arr=df['temp'].values
print(f"temp-median {np.median(temp_arr)}, Std:{np.std(temp_arr)},IQR: {np.percentile(temp_arr,75)-np.percentile(temp_arr,25)}")
```

```python
#Find the 90th percentile for windspeed and precip
print(f"Question 27: 90th percentile: windspeed={np.percentile(df['windspeed'],90):.2f}, precip={np.percentile(df['precip'],90):.2f}")
```

```python
#Normalize the temp column using Min-Max scaling.
df['temp_norm'] = (df['temp'] - df['temp'].min()) / (df['temp'].max() - df['temp'].min())
```

```python
#Calculate the z-score for temperature values and identify outliers (z-score > 3)

# Z-score
df['temp_zscore'] = (df['temp'] - df['temp'].mean()) / df['temp'].std()
outliers = df[np.abs(df['temp_zscore']) > 3]
print(f"Question 29: Number of temp outliers (z>3): {len(outliers)}")

```

```python
#Calculate the covariance matrix between numerical columns: temp, humidity, windspeed, precip.
# Covariance matrix
cov_matrix = df[['temp','humidity','windspeed','precip']].cov()
print("\nQuestion 30: Covariance matrix:")
print(cov_matrix)
```

```python
#Create a new column season based on month (Winter: Dec-Feb, Spring: Mar-May, Summer: Jun-Aug, Fall: Sep-Nov).
def get_season(month):
    if month in [12,1,2]: return "winter"
    if month in [3,4,5]: return "spring"
    if month in [6,7,8]: return "summer"
    else: return "fall"
df['season']=df['month'].apply(get_season)
df
```

```python
#Create a new feature temp_category with categories: "Cold" (< 15°C), "Moderate" (15-30°C), "Hot" (> 30°C).
def temp_category(temp):
    if (temp <15): return "cold"
    if(temp >30): return "hot"
    
    else: return "moderate"
df['tempcategory']=df['temp'].apply(temp_category)
df
```

```python
#Calculate the day of year (dayofyear) and create a new feature day_of_week from the date.
df['day_of_week'] = df.index.dayofweek 
df
```

```python
#Create a lag feature: temp_lag1 with yesterday's temperature.
df['temp_lag1'] = df['temp'].shift(1)
df
```

```python
#Calculate the rate of change in temperature between consecutive days as a new column
df['temp_change'] = df['temp'].diff()
df
```
